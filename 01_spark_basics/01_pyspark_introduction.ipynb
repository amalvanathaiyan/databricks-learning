## Concept
**Apache Spark** is a distributed data processing engine. It splits data and work across many machines and processes them in parallel.

**PySpark** is the Python API for Spark. You write Python, and Spark executes the work on the cluster.

PySpark is used for:

- Data processing and ETL
- Data transformation
- SQL analytics
- Machine learning
- Large-scale data engineering

## Real-world example
A bank receives 10 million transactions per day from Oracle. One machine running Pandas would be slow or run out of memory. Spark splits the data across executors, each processing a part at the same time.

## Why do we use Spark?

Pandas works on a single machine and keeps data in memory. Spark distributes processing across multiple machines.

```
10 million records
        |
        v
  Spark cluster
  +----+----+----+----+
  | E1 | E2 | E3 | E4 |
  +----+----+----+----+
        |
        v
  Process in parallel
```

## Python vs Spark vs PySpark vs Pandas

| Name | What it is |
|---|---|
| Python | A general-purpose programming language |
| Apache Spark | A distributed data processing engine (written in Scala, runs on the JVM) |
| PySpark | The Python API for Spark, so you drive Spark with Python |
| Pandas | A Python library for data on a single machine |

Use Pandas for small data on one machine. Use Spark when data or processing is too big for one machine, or needs to scale.

## Your first DataFrame

A **DataFrame** is a distributed table with named columns. We build one from a Python list.

````python
data = [
    (1, "Amal", "India", 21000),
    (2, "Senthil", "USA", 150000),
    (3, "John", "UK", 18000),
    (4, "Vivek", "UAE", 25000),
]

columns = ["id", "name", "country", "salary"]

df = spark.createDataFrame(data, columns)
````

````python
display(df)
df.printSchema()
df.show()
````

## What Spark is doing internally

```
createDataFrame()
      |
 Builds a plan for the data (no heavy work yet)
      |
show() / display()
      |
 Action: Spark executes the plan and returns rows
```

`printSchema()` only reads metadata, so it is instant.
Actions like `show()` and `display()` are what trigger real execution. We study this properly in the internals folder (lazy evaluation).

## Interview Q&A

1. **What is the difference between Python, PySpark and Apache Spark?**
   Python is a programming language. Apache Spark is a distributed data processing engine. PySpark is the Python API that lets you write Spark applications in Python.

2. **What is SparkSession?**
   The entry point to a Spark application. It is used to create DataFrames, read data, run SQL and configure Spark. In Databricks it is already available as `spark`.

3. **Why Spark instead of Pandas?**
   Pandas runs on one machine and is limited by its memory. Spark distributes data and processing across a cluster, so it scales to much larger datasets.

4. **What is a DataFrame?**
   A distributed collection of data organized into named columns, like a table, with a schema.

5. **Is Spark a database?**
   No. It is a processing engine. It reads data from storage (files, tables) and processes it, but it doesn't store data itself.
