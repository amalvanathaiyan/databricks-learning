## Concept

<div style="background:#E8F1FF;border-left:5px solid #2F6FED;padding:10px 14px;border-radius:6px;color:#1b2a41;">
<b>💡 Spark and PySpark</b><br>
<b>Apache Spark</b> is a distributed data processing engine. It splits data and work across many machines and processes them in parallel.<br><br>
<b>PySpark</b> is the Python API for Spark. You write Python, and Spark executes the work on the cluster.<br><br>
PySpark is used for:
<ul><li>Data processing and ETL</li><li>Data transformation</li><li>SQL analytics</li><li>Machine learning</li><li>Large-scale data engineering</li></ul>
</div>

## Real-world example

<div style="background:#FFF4DB;border-left:5px solid #E5A100;padding:10px 14px;border-radius:6px;color:#4a3500;">
<b>🏦 Real-world example</b><br>
A bank receives 10 million transactions per day from Oracle. One machine running Pandas would be slow or run out of memory. Spark splits the data across executors, each processing a part at the same time.
</div>

## Why do we use Spark?

Pandas works on a single machine and keeps data in memory. Spark distributes processing across multiple machines.

```
10 million records
        |
        v
  Spark cluster
  +----+----+----+----+
  | E1 | E2 | E3 | E4 |
  +----+----+----+----+
        |
        v
  Process in parallel
```

## Python vs Spark vs PySpark vs Pandas

| Name | What it is |
|---|---|
| Python | A general-purpose programming language |
| Apache Spark | A distributed data processing engine (written in Scala, runs on the JVM) |
| PySpark | The Python API for Spark, so you drive Spark with Python |
| Pandas | A Python library for data on a single machine |

<div style="background:#E7F7EE;border-left:5px solid #1E9E5A;padding:10px 14px;border-radius:6px;color:#14391f;">
<b>✅ Rule of thumb</b><br>
Use Pandas for small data on one machine. Use Spark when data or processing is too big for one machine, or needs to scale.
</div>

## 2. Your first DataFrame

<div style="background:#E8F1FF;border-left:5px solid #2F6FED;padding:10px 14px;border-radius:6px;color:#1b2a41;">
<b>💡 What it does</b><br>
Builds a DataFrame from a Python list of tuples and a list of column names.<br><br>
<b>Why it matters</b><br>A <b>DataFrame</b> is a distributed table with named columns. It is the object you will use in almost every lesson. Lesson 3 covers it in depth.
</div>
<br>
<div style="background:#E7F7EE;border-left:5px solid #1E9E5A;padding:10px 14px;border-radius:6px;color:#14391f;">
<b>✅ Expected output</b><br>
No output yet. Creating the DataFrame only builds a plan.
</div>

````python
data = [
    (1, "Amal", "India", 21000),
    (2, "Senthil", "USA", 150000),
    (3, "John", "UK", 18000),
    (4, "Vivek", "UAE", 25000),
]

columns = ["id", "name", "country", "salary"]

df = spark.createDataFrame(data, columns)
````

## 3. Look at the data

<div style="background:#E8F1FF;border-left:5px solid #2F6FED;padding:10px 14px;border-radius:6px;color:#1b2a41;">
<b>💡 What it does</b><br>
Shows the DataFrame as an interactive table, prints its schema, and prints it as text.<br><br>
<b>Why it matters</b><br><code>display()</code> and <code>show()</code> are <b>actions</b>: they make Spark do real work. <code>printSchema()</code> only reads metadata.
</div>
<br>
<div style="background:#E7F7EE;border-left:5px solid #1E9E5A;padding:10px 14px;border-radius:6px;color:#14391f;">
<b>✅ Expected output</b><br>
An interactive table, a schema tree (<code>id: long</code>, <code>name: string</code>, <code>country: string</code>, <code>salary: long</code>), then a text table with 4 rows.
</div>

````python
display(df)
df.printSchema()
df.show()
````

## What Spark is doing internally

```
createDataFrame()
      |
 Builds a plan for the data (no heavy work yet)
      |
show() / display()
      |
 Action: Spark executes the plan and returns rows
```

<span style="background:#1E9E5A;color:white;padding:2px 10px;border-radius:12px;font-size:12px;">Metadata only</span> `printSchema()` only reads metadata, so it is instant.

<span style="background:#D64545;color:white;padding:2px 10px;border-radius:12px;font-size:12px;">Action</span> `show()` and `display()` trigger real execution.

We study this properly in the internals folder (lazy evaluation).

## Interview Q&A

<div style="background:#F1E9FF;border-left:5px solid #7A4DD8;padding:10px 14px;border-radius:6px;color:#2d1a57;">
<b>🎤 1. What is the difference between Python, PySpark and Apache Spark?</b><br>
Python is a programming language. Apache Spark is a distributed data processing engine. PySpark is the Python API that lets you write Spark applications in Python.<br><br>

<b>🎤 2. What is SparkSession?</b><br>
The entry point to a Spark application. It is used to create DataFrames, read data, run SQL and configure Spark. In Databricks it is already available as <code>spark</code>.<br><br>

<b>🎤 3. Why Spark instead of Pandas?</b><br>
Pandas runs on one machine and is limited by its memory. Spark distributes data and processing across a cluster, so it scales to much larger datasets.<br><br>

<b>🎤 4. What is a DataFrame?</b><br>
A distributed collection of data organized into named columns, like a table, with a schema.<br><br>

<b>🎤 5. Is Spark a database?</b><br>
No. It is a processing engine. It reads data from storage (files, tables) and processes it, but it doesn't store data itself.
</div>
