## Goal
Understand the Databricks workspace layout and run your first commands.

## Why it matters
Everything you build later (notebooks, tables, jobs, dashboards) lives somewhere in the workspace. Knowing where things live saves time and makes your documentation clear to others.

## Concept
A Databricks **workspace** is your web environment for data work. It has:

- **Workspace**: folders, notebooks, files and Git folders (this is where your repo lives)
- **Catalog**: the data you can query (catalogs, schemas, tables, volumes)
- **Compute**: where your code runs (serverless in Free Edition)
- **Jobs & Pipelines**: scheduled and orchestrated work
- **SQL editor, dashboards, Genie**: analytics and BI tools

## Real-world example
A data engineering team keeps notebooks in Git folders, tables in Unity Catalog, and runs them nightly through Jobs. This workspace is a small version of that setup.

## UI map

| Sidebar item | What it is for |

# 1. Workspace basics

## 1.1 Hello Databricks and Spark version

**What it does:** prints text and shows the Spark version running on your compute.
**Why it matters:** confirms your notebook is attached to working compute, and the version tells you which Spark features are available.
**Expected output:** `Hello Databricks` and a version string.

```python
print("Hello Databricks")
print(spark.version)
```

## 1.2 Who and where am I?

**What it does:** asks Spark SQL for the current user, catalog and schema.
**Why it matters:** every table you create or query lives in a catalog and schema. Knowing your defaults avoids creating tables in the wrong place.
**Expected output:** one row with your email, the current catalog (often `workspace`) and schema (often `default`).

```python
display(spark.sql("SELECT current_user() AS me, current_catalog() AS catalog, current_schema() AS schema"))
```

## 1.3 What data can I see?

**What it does:** lists the catalogs you can access, then the schemas in your current catalog.
**Why it matters:** Unity Catalog organizes data as catalog, schema, table. This is how you explore it from code.
**Expected output:** a list of catalogs, then a list of schemas.

```python
display(spark.sql("SHOW CATALOGS"))
display(spark.sql("SHOW SCHEMAS"))

print("** Checking with lowercase **")
display(spark.sql("show catalogs")) # works
display(spark.sql("show schemas")) #works
print("** Yes It Works **")
```

## Interview Q&A

1. **What is a Databricks workspace?**
   The web environment where you organize notebooks, files, compute, jobs and data assets for a team or project.

2. **What is the difference between the workspace and the catalog?**
   The workspace holds code and files (notebooks, folders). The catalog (Unity Catalog) holds governed data objects like tables and volumes.

3. **Where does your code actually run?**
   On compute attached to the notebook, which is serverless in Free Edition.

4. **What is a Git folder?**
   A workspace folder linked to a Git repo, so you can commit, push and pull notebooks directly from Databricks.
