## Concept

<div style="background:#E8F1FF;border-left:5px solid #2F6FED;padding:10px 14px;border-radius:6px;color:#1b2a41;">
<b>💡 What is a workspace?</b><br>
A Databricks <b>workspace</b> is your web environment for data work. It has:
<ul><li><b>Workspace</b>: folders, notebooks, files and Git folders (this is where your repo lives)</li><li><b>Catalog</b>: the data you can query (catalogs, schemas, tables, volumes)</li><li><b>Compute</b>: where your code runs (serverless in Free Edition)</li><li><b>Jobs & Pipelines</b>: scheduled and orchestrated work</li><li><b>SQL editor, dashboards, Genie</b>: analytics and BI tools</li></ul>
</div>

## Real-world example

<div style="background:#FFF4DB;border-left:5px solid #E5A100;padding:10px 14px;border-radius:6px;color:#4a3500;">
<b>🏦 Real-world example</b><br>
A data engineering team keeps notebooks in Git folders, tables in Unity Catalog, and runs them nightly through Jobs. This workspace is a small version of that setup.

## UI map

| Sidebar item | What it is for |
|---|---|
| Workspace | Notebooks, folders, Git folders, shared files |
| Catalog | Browse catalogs, schemas, tables, volumes |
| Jobs & Pipelines | Schedule and monitor workloads |
| Compute | See compute resources (serverless in Free Edition) |
| Recents | Quickly reopen what you worked on |
| Marketplace | Datasets and solutions shared by others |

<div style="background:#E7F7EE;border-left:5px solid #1E9E5A;padding:10px 14px;border-radius:6px;color:#14391f;">
<b>✅ Workspace vs catalog</b><br>
The <b>workspace</b> holds code (notebooks, files). The <b>catalog</b> holds governed data (tables, views, volumes, functions). A notebook in the workspace reads and writes data in the catalog.
</div>

## The Unity Catalog hierarchy

```
metastore (one per region, managed by Databricks)
└── catalog          e.g. workspace          <- Free Edition gives you 'workspace'
    └── schema       e.g. default
        ├── table    e.g. workspace.default.sales
        ├── view
        ├── volume   e.g. /Volumes/workspace/default/raw_files   (files)
        └── function
```

Every data object has a three-part name: `catalog.schema.object`. That hierarchy is covered in depth in `06_unity_catalog`.

## 2. Who and where am I?

<div style="background:#E8F1FF;border-left:5px solid #2F6FED;padding:10px 14px;border-radius:6px;color:#1b2a41;">
<b>💡 What it does</b><br>
Asks Spark SQL for the current user, catalog and schema.<br><br>
<b>Why it matters</b><br>Every table you create or query lives in a catalog and schema. Knowing your defaults avoids creating tables in the wrong place.
</div>
<br>
<div style="background:#E7F7EE;border-left:5px solid #1E9E5A;padding:10px 14px;border-radius:6px;color:#14391f;">
<b>✅ Expected output</b><br>
One row with your email, the current catalog (often <code>workspace</code>) and schema (often <code>default</code>).
</div>

````python
display(spark.sql("SELECT current_user() AS me, current_catalog() AS catalog, current_schema() AS schema"))
````

## 3. What data can I see?

<div style="background:#E8F1FF;border-left:5px solid #2F6FED;padding:10px 14px;border-radius:6px;color:#1b2a41;">
<b>💡 What it does</b><br>
Lists the catalogs you can access, then the schemas in your current catalog.<br><br>
<b>Why it matters</b><br>Unity Catalog organizes data as catalog, schema, table. This is how you explore it from code. SQL keywords are case-insensitive, which the second half of the cell checks.
</div>
<br>
<div style="background:#E7F7EE;border-left:5px solid #1E9E5A;padding:10px 14px;border-radius:6px;color:#14391f;">
<b>✅ Expected output</b><br>
A list of catalogs, then a list of schemas, twice (upper and lower case), then <code>** Yes It Works **</code>.
</div>

````python
display(spark.sql("SHOW CATALOGS"))
display(spark.sql("SHOW SCHEMAS"))

print("** Checking with lowercase **")
display(spark.sql("show catalogs")) # works
display(spark.sql("show schemas")) #works
print("** Yes It Works **")
````

## 4. Explore a sample dataset

<div style="background:#E8F1FF;border-left:5px solid #2F6FED;padding:10px 14px;border-radius:6px;color:#1b2a41;">
<b>💡 What it does</b><br>
Lists the tables in the read-only <code>samples</code> catalog that Databricks provides, and reads a few rows.<br><br>
<b>Why it matters</b><br><code>samples</code> is a quick source of realistic data for experiments. Fully qualified three-part names (<code>catalog.schema.table</code>) work from anywhere, whatever your current catalog is.
</div>
<br>
<div style="background:#FFF4DB;border-left:5px solid #E5A100;padding:10px 14px;border-radius:6px;color:#4a3500;">
<b>🆓 Free Edition note</b><br>
If <code>samples</code> isn't visible in your workspace, skip this cell and note it in the troubleshooting log.
</div>
<br>
<div style="background:#E7F7EE;border-left:5px solid #1E9E5A;padding:10px 14px;border-radius:6px;color:#14391f;">
<b>✅ Expected output</b><br>
A list of tables in <code>samples.nyctaxi</code> (such as <code>trips</code>), then 5 taxi trips.
</div>

````python
display(spark.sql("SHOW TABLES IN samples.nyctaxi"))
display(spark.table("samples.nyctaxi.trips").limit(5))
````

## Interview Q&A

<div style="background:#F1E9FF;border-left:5px solid #7A4DD8;padding:10px 14px;border-radius:6px;color:#2d1a57;">
<b>🎤 1. What is a Databricks workspace?</b><br>
The web environment where you organize notebooks, files, compute, jobs and data assets for a team or project.<br><br>

<b>🎤 2. What is the difference between the workspace and the catalog?</b><br>
The workspace holds code and files (notebooks, folders). The catalog (Unity Catalog) holds governed data objects like tables and volumes.<br><br>

<b>🎤 3. Where does your code actually run?</b><br>
On compute attached to the notebook, which is serverless in Free Edition.<br><br>

<b>🎤 4. What is a Git folder?</b><br>
A workspace folder linked to a Git repo, so you can commit, push and pull notebooks directly from Databricks.<br><br>

<b>🎤 5. What is the three-level namespace in Unity Catalog?</b><br>
<code>catalog.schema.object</code>, for example <code>workspace.default.sales</code>. Every table, view, volume and function is addressed that way.
</div>
