# 00_04 Git Workflow

## Goal
Use Git from inside Databricks: commit, push, pull, and work on a branch.

## Why it matters
Notebooks are code. Version control gives you history, safe experiments through branches, and a public portfolio on GitHub. Git workflow questions also come up in data engineering interviews.

## Concept
A **Git folder** is a workspace folder linked to a GitHub repo. Typical flow:

1. Pull the latest changes
2. Create a branch for your work
3. Edit notebooks
4. Commit with a clear message
5. Push the branch
6. Open a pull request on GitHub and merge it
7. Pull `main` again

## Real-world example
A team never edits `main` directly. Each engineer works on a feature branch, opens a pull request, gets a review, and merges. Production jobs run from the merged code.

## Key terms

| Term | Meaning |
|---|---|
| Repository | The project and its full history on GitHub |
| Commit | A saved snapshot with a message |
| Push | Send your commits to GitHub |
| Pull | Bring GitHub changes into your workspace |
| Branch | A separate line of work, like `feature/add-lesson-1` |
| Pull request (PR) | A request to merge a branch into `main` |
| Merge conflict | Two changes to the same lines that Git can't combine automatically |

| Problem | Fix |
|---|---|
| Push fails with an authentication error | Reconnect GitHub in User Settings → Linked accounts, using a personal access token with `repo` scope |
| Commit button is disabled | Select at least one changed file and enter a commit message |
| | |

## Commit message habits

- Short and specific: `add 17 joins lesson`, not `update`
- One lesson or change per commit
- Commit after every completed notebook

## What to keep out of Git

- Passwords, tokens and secrets (use secret scopes instead)
- Large data files

## Interview Q&A

1. **Why use Git folders in Databricks?**
   They link workspace notebooks to a repo so you get version control, branching and collaboration without leaving Databricks.

2. **What is the difference between commit and push?**
   A commit saves a snapshot in your local history. A push sends those commits to the remote repo.

3. **Why work on branches instead of `main`?**
   Branches isolate changes, allow review through pull requests, and keep `main` stable for production.

4. **What is a merge conflict and how do you handle it?**
   It happens when two branches change the same lines. You open the file, choose or combine the correct content, then commit the resolution.

5. **What should never be committed?**
   Secrets, credentials and large data files.
