## Concept

<div style="background:#E8F1FF;border-left:5px solid #2F6FED;padding:10px 14px;border-radius:6px;color:#1b2a41;">
<b>💡 Git folders</b><br>
A <b>Git folder</b> is a workspace folder linked to a GitHub repo. Typical flow:
<ol><li>Pull the latest changes</li><li>Create a branch for your work</li><li>Edit notebooks</li><li>Commit with a clear message</li><li>Push the branch</li><li>Open a pull request on GitHub and merge it</li><li>Pull <code>main</code> again</li></ol>
</div>

## Real-world example

<div style="background:#FFF4DB;border-left:5px solid #E5A100;padding:10px 14px;border-radius:6px;color:#4a3500;">
<b>🏦 Real-world example</b><br>
A team never edits <code>main</code> directly. Each engineer works on a feature branch, opens a pull request, gets a review, and merges. Production jobs run from the merged code.

## Key terms

| Term | Meaning |
|---|---|
| Repository | The project and its full history on GitHub |
| Commit | A saved snapshot with a message |
| Push | Send your commits to GitHub |
| Pull | Bring GitHub changes into your workspace |
| Branch | A separate line of work, like `feature/add-lesson-1` |
| Pull request (PR) | A request to merge a branch into `main` |
| Merge conflict | Two changes to the same lines that Git can't combine automatically |

## Where to click in Databricks

| Action | Where |
|---|---|
| Link GitHub | Settings → Linked accounts → Git integration (personal access token or GitHub app) |
| Create a Git folder | Workspace → Create → Git folder → paste the repo URL |
| Open the Git dialog | The branch name button next to the folder or notebook title |
| Switch / create branch | Git dialog → branch dropdown |
| Commit and push | Git dialog → select files → message → **Commit & Push** |
| Pull | Git dialog → **Pull** |

## How this scales: from Git folders to CI/CD

```
Developer workspace (Git folder, feature branch)
        |  commit & push
        v
GitHub  --pull request + review-->  main
        |
        v
CI/CD (GitHub Actions / Azure DevOps)
        |  runs tests, then deploys with Databricks Asset Bundles
        v
Staging workspace  -->  Production workspace (jobs run from main)
```

**Databricks Asset Bundles** describe jobs, pipelines and their settings as YAML files in the repo, so the same code can be deployed to dev, staging and prod with one command. They are covered in `09_azure_databricks` (CI/CD).

## Commit message habits

- Short and specific: `add 17 joins lesson`, not `update`
- One lesson or change per commit
- Commit after every completed notebook

## What to keep out of Git

- Passwords, tokens and secrets (use secret scopes instead)
- Large data files

<div style="background:#FDEAEA;border-left:5px solid #D64545;padding:10px 14px;border-radius:6px;color:#4a1414;">
<b>⛔ Never commit credentials</b><br>
A token pushed to GitHub, even once, should be treated as leaked: revoke it and create a new one. Store secrets in a password manager or a Databricks secret scope, never in notebooks or text files in the repo.
</div>

## Common mistakes and troubleshooting log

<div style="background:#FDEAEA;border-left:5px solid #D64545;padding:10px 14px;border-radius:6px;color:#4a1414;">
<b>⛔ Problem: push fails with an authentication error</b><br>
Reconnect GitHub in Settings → Linked accounts, using a personal access token with <code>repo</code> scope.<br><br>
<b>⛔ Problem: Commit button is disabled</b><br>Select at least one changed file and enter a commit message.<br><br>
<b>⛔ Problem: pull reports a conflict</b><br>Someone changed the same lines. Resolve it in the Git dialog (or on GitHub), then commit the resolution.<br><br>
<b>⛔ Problem: notebook outputs make diffs huge</b><br>Clear outputs before committing, or adjust the Git folder's output-commit setting.<br><br>

## Interview Q&A

<div style="background:#F1E9FF;border-left:5px solid #7A4DD8;padding:10px 14px;border-radius:6px;color:#2d1a57;">
<b>🎤 1. Why use Git folders in Databricks?</b><br>
They link workspace notebooks to a repo so you get version control, branching and collaboration without leaving Databricks.<br><br>

<b>🎤 2. What is the difference between commit and push?</b><br>
A commit saves a snapshot in your local history. A push sends those commits to the remote repo.<br><br>

<b>🎤 3. Why work on branches instead of <code>main</code>?</b><br>
Branches isolate changes, allow review through pull requests, and keep <code>main</code> stable for production.<br><br>

<b>🎤 4. What is a merge conflict and how do you handle it?</b><br>
It happens when two branches change the same lines. You open the file, choose or combine the correct content, then commit the resolution.<br><br>

<b>🎤 5. What should never be committed?</b><br>
Secrets, credentials and large data files.<br><br>

<b>🎤 6. How does code get from a Git folder to production on Databricks?</b><br>
Changes are merged to <code>main</code> through pull requests, then a CI/CD pipeline runs tests and deploys jobs and pipelines to staging and production workspaces, typically with Databricks Asset Bundles.
</div>
