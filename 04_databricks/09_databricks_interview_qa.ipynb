## Core explanations

<div style="background:#F1E9FF;border-left:5px solid #7A4DD8;padding:10px 14px;border-radius:6px;color:#2d1a57;">
<b>🎤 1. Describe the Databricks platform architecture.</b><br>
The control plane, managed by Databricks, hosts the web UI, notebooks, jobs and APIs. The compute plane runs the workloads: serverless compute in Databricks' account, or classic clusters in your cloud account. Data stays in your cloud storage as Delta tables and volumes, governed by Unity Catalog, which manages permissions, lineage and audit across workspaces. On top sit the tools: notebooks, jobs and pipelines, SQL warehouses, dashboards and Genie.<br><br>

<b>🎤 2. How do you choose compute for a workload?</b><br>
Serverless notebooks for interactive work, serverless jobs or job clusters for scheduled pipelines, SQL warehouses for SQL, dashboards and BI, and classic clusters only when I need specific hardware, libraries or APIs. Production never runs on always-on all-purpose clusters, because job compute is cheaper and isolated.<br><br>

<b>🎤 3. How do you structure a Databricks project for production?</b><br>
A Git folder with thin notebooks as entry points, transformation logic in Python modules with unit tests, configuration through parameters (widgets and job parameters), secrets in secret scopes, data in Unity Catalog tables and volumes, and deployment with Asset Bundles through CI/CD to dev, test and prod.<br><br>

<b>🎤 4. How do you handle parameters and secrets in notebooks?</b><br>
Parameters come from widgets, filled by job parameters, and are validated at the top into a config object. In SQL they're passed as parameter markers and <code>IDENTIFIER()</code>. Secrets live in secret scopes and are read with <code>dbutils.secrets.get</code>, redacted in output, with ACLs controlling access. For storage and databases I prefer Unity Catalog credentials and connections.<br><br>

<b>🎤 5. How do you monitor a Databricks platform?</b><br>
Job notifications and run history for pipelines, query history and profiles for SQL performance, freshness and volume checks from Delta history plus data quality checks for data, and system tables for cost, usage, job runs and audit, with dashboards and alerts on top.
</div>

## Rapid-fire questions

<div style="background:#F1E9FF;border-left:5px solid #7A4DD8;padding:10px 14px;border-radius:6px;color:#2d1a57;">
<b>🎤 1. Control plane vs compute plane?</b><br>
The control plane is the Databricks-managed UI, APIs and job scheduler. The compute plane is where data is processed: serverless or your cloud account's clusters.<br><br>

<b>🎤 2. What is a Git folder?</b><br>
A workspace folder linked to a remote Git repository, for branches, commits and pull requests in Databricks.<br><br>

<b>🎤 3. <code>%run</code> vs <code>dbutils.notebook.run</code>?</b><br>
<code>%run</code> runs a notebook inline in the same context. <code>dbutils.notebook.run</code> runs it separately with parameters and returns a value.<br><br>

<b>🎤 4. How do job tasks share values?</b><br>
Through <code>dbutils.jobs.taskValues</code>, or tables for larger data.<br><br>

<b>🎤 5. Widget value types?</b><br>
Always strings.<br><br>

<b>🎤 6. What does <code>dbutils.secrets.get</code> show when printed?</b><br>
<code>[REDACTED]</code>.<br><br>

<b>🎤 7. Size vs scaling on a SQL warehouse?</b><br>
Size is for query complexity. Scaling (number of clusters) is for concurrency.<br><br>

<b>🎤 8. What is the Statement Execution API?</b><br>
A REST API for running SQL on a SQL warehouse from applications, without Spark.<br><br>

<b>🎤 9. What is Genie?</b><br>
Natural-language questions over Unity Catalog tables, answered with generated SQL on a warehouse.<br><br>

<b>🎤 10. What are system tables?</b><br>
Tables in the <code>system</code> catalog with billing, query history, job runs, audit and lineage data.<br><br>

<b>🎤 11. What is an instance pool?</b><br>
Idle, warm VMs that speed up classic cluster starts.<br><br>

<b>🎤 12. What is a cluster policy?</b><br>
Admin rules that limit and default cluster settings, for cost and governance.

## Scenario questions

<div style="background:#F1E9FF;border-left:5px solid #7A4DD8;padding:10px 14px;border-radius:6px;color:#2d1a57;">
<b>🎤 1. The monthly Databricks bill doubled. How do you investigate?</b><br>
Query <code>system.billing.usage</code> by day, product and job or warehouse to find what grew, and join list prices for cost. Then look at that workload's runs, such as schedule changes, longer durations or a warehouse that never stops. Fix the cause and add a cost dashboard or alert.<br><br>

<b>🎤 2. A notebook works for you but fails as a scheduled job. What do you check?</b><br>
That parameters and widgets are defined with defaults, that the job's identity has permissions on the tables, volumes and secrets, that it doesn't depend on temp views or state from another notebook, the compute and environment differences (libraries, serverless restrictions), and that paths aren't relative to a personal folder.<br><br>

<b>🎤 3. An API token was found in a notebook in Git. What do you do?</b><br>
Revoke and rotate the token immediately, store the new one in a secret scope with tight ACLs, update the code to <code>dbutils.secrets.get</code>, and add secret scanning to the repository.<br><br>

<b>🎤 4. Analysts complain the dashboard is slow every Monday morning.</b><br>
Check Query History for queuing versus long execution. If queries queue, increase the warehouse's max clusters. If single queries are slow, look at the profiles and pre-aggregate or cluster the Gold tables, and check that results can be cached.<br><br>

<b>🎤 5. How would you let 200 business users explore sales data without writing SQL?</b><br>
Build clean, documented Gold tables and views with a clear grain, then publish an AI/BI dashboard for the standard KPIs and a Genie space with instructions and example queries for ad hoc questions, all governed by Unity Catalog permissions.
</div>
